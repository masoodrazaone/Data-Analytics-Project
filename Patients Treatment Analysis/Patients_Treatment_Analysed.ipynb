```python
import re
def find_contact_details(text: str) -> tuple:
    # it the value is NaN, then return it
    if pd.isna(text):
        return np.nan

    # create the phone number pattern
    phone_number_pattern = re.compile(r"(\+[\d]{1,3}\s)?(\(?[\d]{3}\)?\s?-?[\d]{3}\s?-?[\d]{4})")
    # find the phone number from the value/text, as a result we will get a list
    phone_number  = re.findall(phone_number_pattern, text)

    # if length is 0, then the regex can't find any ph number, then define with NaN
    if len(phone_number) <= 0:
        phone_number = np.nan
    # if the country code is attached with the ph number, for that case, the first
    # element will be the country code and the 2nd element will be the actual ph
    # number. So, get that ph number
    elif len(phone_number) >= 2:
        phone_number = phone_number[1]
    # else, we will get the ph number. Grab it.
    else:
        phone_number = phone_number[0]

    # if we found the ph number (with/without country code), then remove that part from the actual value.
    # after removing the ph number, the remaining string might be the email address.
    possible_email_add = re.sub(phone_number_pattern, "", text).strip()

    # then return the ph number and the email address
    return phone_number, possible_email_add
```

```python
patients_df['phone'] = patients_df["contact"].apply(lambda x: find_contact_details(x)).apply(lambda x:x[0])
patients_df['email'] = patients_df["contact"].apply(lambda x: find_contact_details(x)).apply(lambda x:x[1])
```

**Patient Dataset Issues**

Text formatting inconsistencies across 5 location/name fields

Zip code formatting inconsistencies

Duplicate records detected and removed

Mixed formatting in contact information

**Treatment Dataset Issues**

Treatment data split across 2 datasets

**Adverse Reaction Dataset Issues**

Large number of missing values representing absence of reactions

## 4. Data Cleaning Process
**4.1 Datatype Corrections**

Converted birthdate from object → datetime.

Standardized categorical variables.

Converted dosage and measurement fields into numeric format.

**4.2 Data Standardization**

Standardized capitalization across names, city, and state columns.

Corrected zip code formatting for all records.

Removed duplicate entries ensuring unique patient records (499 final patients).

**4.3 Tidy Data Transformation**

The treatments dataset was reshaped using melt():

Converted wide format → long format.

Reduced structural redundancy.

Ensured each observation represents a single patient-treatment record.

This step transformed the dataset into a tidy data structure, improving usability for analysis.

**4.4 Dataset Integration**

Combined treatments and treatments_cut datasets.

Merged treatment data with patient information.

Integrated adverse reaction records.

Final master dataset created successfully.
